```python
#Grade Calcuator
student_name= input("Enter the student name: ")

student_record={}

while True:
   subject=input("Enter subject name:")
   if subject.lower()=="done":
       break
   try: 
    student_grade= int(input("Enter Student Grade:"))
   except ValueError:
       print("invalid input. Please Try again")
       continue

   if 100>=student_grade>=80:
        letter= "A"
   elif 80>student_grade>=75:
        letter="B+"
   elif 75>student_grade>=70:
        letter="B"    
   elif 70>student_grade>=60:
        letter="C" 
   elif 60>student_grade>=50:
        letter="D"   
   elif student_grade<50:
        letter="F"   
        
   student_record[subject]=letter
  
   
# Print the final report card
print("\n" + "="*30)
print(f"Report Card : {student_name.upper()}")
print("="*30)
print(f"{'SUBJECT': <20} | {'GRADE':<5}")
print("="*30)

for subject,grade in student_record.items():
    print(f"{subject:<20} | {grade:<5}")

print("="*30)
        
    

```

```python
import random

def number_guessing_game():
   
    LOWER_BOUND = 1
    UPPER_BOUND = 100
    secret_number = random.randint(LOWER_BOUND, UPPER_BOUND)
    guesses_taken = 0

    print("=============================")
    print("Let the Games Begin!")
    print(f"I'm thinking of a number between {LOWER_BOUND} and {UPPER_BOUND}.")
    print("Take your time, lets see if you can win!")
    print("=============================\n")

    
    while True: 
        try:
            guess = int(input("Enter your guess: "))
            guesses_taken += 1 # Increment attempts count
        except ValueError:
            print("Invalid input! Please enter a whole number (integer).")
            # Adjust counter back since this was not a valid guess attempt
            guesses_taken -= 1
            continue # Restart the loop iteration without running comparison logic

       
        if guess < secret_number:
            print("⬆Are you serious? Try guessing higher.")
        elif guess > secret_number:
            print("⬇Are you HIGH, coz your answer is!! Try guessing lower.")
        else:
            # Success! Break the loop and end the game.
            break

   
    print("\n*******************************")
    print(f"CONGRATULATIONS! You guessed the number {secret_number}!")
    print(f"You won in {guesses_taken} attempts!")
    print("Thanks for playing! Come back soon.")
    print("*******************************")

if __name__ == "__main__":
    number_guessing_game()
```

```python
import math 

def check_prime(number):
    """Checks if a given whole number is prime."""
    
   
    if number <= 1:
        return False
    

    if number == 2:
        return True

    
    if number % 2 == 0:
        return False

    
    limit = int(math.sqrt(number)) + 1
    for i in range(3, limit, 2):
        if number % i == 0:
            return False 

    
    return True



if __name__ == "__main__":
    print("=============================")
    print("Prime Number Checker")
    print("=============================")
    
    while True:
        user_input = input("\nEnter a positive whole number to check (or 'q' to quit): ")
        
        if user_input.lower() == 'q':
            print("\nExiting checker. Goodbye!")
            break
        
        try:
            number = int(user_input)
            result = check_prime(number)

           
            if result is True:
                print(f"\n{number} IS a prime number.")
            else:
                print(f"\n{number} is NOT a prime number.")

        except ValueError:
            print("\n Invalid input. Please enter a valid whole integer or 'q'.")

```

```python

matrix = [
    [1, 2, 3],  
    [4, 5, 6],  
    [7, 8, 9]   
]

total_sum = 0
ROWS = 3
COLS = 3

print("--- Calculating Sum using Nested Loops ---")


for i in range(ROWS):
    #
    for j in range(COLS):
        element = matrix[i][j]
        total_sum += element
        print(f"Processing element at ({i}, {j}): {element}")

print("\n$$$$$$$$$$$$$$$$$$$$")
print(f"The total sum of the 3x3 matrix is: {total_sum}")
```
